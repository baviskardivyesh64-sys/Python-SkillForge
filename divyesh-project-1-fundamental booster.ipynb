```python
print("===== TYPE CASTING PROJECT =====")

name = input("Enter your name: ")
age = int(input("Enter your age: "))
height = float(input("Enter your height: "))

print()
print("===== ORIGINAL DATA =====")

print("Name:", name)
print("Age:", age)
print("Height:", height)

print()
print("===== TYPE CASTING =====")

age_float = float(age)
height_int = int(height)

print("Age as float:", age_float)
print("Height as integer:", height_int)

print()
print("===== FINAL OUTPUT =====")

print("Name:", name)
print("Age:", age)
print("Height:", height)
```

```python
print("===== AGE CALCULATOR =====")

name = input("Enter your name: ")
age = int(input("Enter your age: "))

print()
print("===== CALCULATIONS =====")

print("Your age is:", age)

print("Age after 5 years:", age + 5)

print("Age 5 years ago:", age - 5)

print("Double of your age:", age * 2)

print("Half of your age:", age / 2)

birth_year = 2026 - age

print("Your approximate birth year is:", birth_year)

print()
print("Hello", name + "!")
print("Your current age is", age)
print("Your approximate birth year is", birth_year)
```

```python
print("===== ID AND TYPE PROJECT =====")

name = "Divyesh"
age = 18
height = 5.7
student = True

print()
print("===== VARIABLE DETAILS =====")

print("Name:", name)
print("Type of name:", type(name))
print("ID of name:", id(name))

print()

print("Age:", age)
print("Type of age:", type(age))
print("ID of age:", id(age))

print()

print("Height:", height)
print("Type of height:", type(height))
print("ID of height:", id(height))

print()

print("Student:", student)
print("Type of student:", type(student))
print("ID of student:", id(student))

print()
print("===== SUMMARY =====")

print("Name =", name, "- Type:", type(name), "- ID:", id(name))
print("Age =", age, "- Type:", type(age), "- ID:", id(age))
print("Height =", height, "- Type:", type(height), "- ID:", id(height))
print("Student =", student, "- Type:", type(student), "- ID:", id(student))
```
